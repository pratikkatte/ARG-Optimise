```python
import sys
from pathlib import Path

import pandas as pd
from scripts.plot_point_accuracy import load_truth_tracks

VALIDATION_DIR = Path(".").resolve()
TRUTH_DIR = VALIDATION_DIR / "tcoalmsp" / "rep0"
TRUTH_PREFIX = "sim_l1mb_0"
NE = 10000

truth_tracks = load_truth_tracks(TRUTH_DIR, TRUTH_PREFIX, NE)
print(f"Loaded {len(truth_tracks)} sample-pair truth tracks from {TRUTH_DIR}")
```

```python
rows = []
for (s1, s2), intervals in sorted(truth_tracks.items()):
    pair = f"{s1}-{s2}"
    for interval in intervals:
        rows.append(
            {
                "pair": pair,
                "left": interval.left,
                "right": interval.right,
                "tcoal_2ne": interval.tcoal_2ne,
                "tcoal_raw": interval.tcoal_2ne * 2 * NE,
            }
        )

df = pd.DataFrame(rows).sort_values(["pair", "left"]).reset_index(drop=True)
print(f"Total intervals: {len(df)}")
df
```

```python
unique_raw = sorted(df["tcoal_raw"].unique())
unique_2ne = sorted(df["tcoal_2ne"].unique())

print(f"Globally unique raw coalescence times: {len(unique_raw)}")
print(unique_raw)
print()
print(f"Globally unique t/(2Ne) coalescence times: {len(unique_2ne)}")
print(unique_2ne)
print()

pair_stats = (
    df.groupby("pair")
    .agg(
        n_intervals=("left", "size"),
        n_unique_times=("tcoal_raw", "nunique"),
        min_tcoal_raw=("tcoal_raw", "min"),
        max_tcoal_raw=("tcoal_raw", "max"),
    )
    .reset_index()
)
print("Per-pair summary:")
display(pair_stats)

selected_pair = "0-1"
pair_times = sorted(df.loc[df["pair"] == selected_pair, "tcoal_raw"].unique())
print(f"\nUnique raw times for pair {selected_pair} ({len(pair_times)} values):")
print(pair_times)
```

### Inferred trees (`example.trees`)

**Time units:** msprime truth stores node/tmrca times in **generations** (raw values ~10³–10⁵), then validation divides by `2*Ne` to get **t/(2Ne)**.

The GFN export in `env.py` now writes learned node times in **generations** and sets `time_units="generations"`, so inferred trees should be divided by `2*Ne` exactly like truth.

```python
import tskit
from scripts.plot_point_accuracy import TruthInterval

INFERRED_TREES = (VALIDATION_DIR.parent / "example_time.trees").resolve()
pairs = list(truth_tracks.keys())


def load_inferred_tracks_from_trees(trees_path, pairs):
    """GFN-inferred trees store tmrca in generations, matching msprime truth."""
    ts = tskit.load(str(trees_path))
    scale = 2.0 * NE
    samples = ts.samples()
    tracks = {pair: [] for pair in pairs}
    for tree in ts.trees():
        left, right = tree.interval
        if right <= left:
            continue
        for pair in pairs:
            tcoal_2ne = float(tree.tmrca(samples[pair[0]], samples[pair[1]]) / scale)
            tracks[pair].append(
                TruthInterval(left=float(left), right=float(right), tcoal_2ne=tcoal_2ne)
            )
    return tracks


inferred_tracks = load_inferred_tracks_from_trees(INFERRED_TREES, pairs)
print(f"Loaded {len(inferred_tracks)} sample-pair inferred tracks from {INFERRED_TREES}")
```

```python
inferred_rows = []
for (s1, s2), intervals in sorted(inferred_tracks.items()):
    pair = f"{s1}-{s2}"
    for interval in intervals:
        inferred_rows.append(
            {
                "pair": pair,
                "left": interval.left,
                "right": interval.right,
                "tcoal_2ne": interval.tcoal_2ne,
                "tcoal_raw": interval.tcoal_2ne * 2 * NE,
            }
        )

inferred_df = (
    pd.DataFrame(inferred_rows).sort_values(["pair", "left"]).reset_index(drop=True)
)
print(f"Total intervals: {len(inferred_df)}")
inferred_df
```

```python
inferred_unique_raw = sorted(inferred_df["tcoal_raw"].unique())
inferred_unique_2ne = sorted(inferred_df["tcoal_2ne"].unique())

print(f"Globally unique raw coalescence times: {len(inferred_unique_raw)}")
print(inferred_unique_raw)
print()
print(f"Globally unique t/(2Ne) coalescence times: {len(inferred_unique_2ne)}")
print(inferred_unique_2ne)
print()

inferred_pair_stats = (
    inferred_df.groupby("pair")
    .agg(
        n_intervals=("left", "size"),
        n_unique_times=("tcoal_raw", "nunique"),
        min_tcoal_raw=("tcoal_raw", "min"),
        max_tcoal_raw=("tcoal_raw", "max"),
    )
    .reset_index()
)
print("Per-pair summary (inferred):")
display(inferred_pair_stats)

selected_pair = "0-1"
inferred_pair_times = sorted(
    inferred_df.loc[inferred_df["pair"] == selected_pair, "tcoal_raw"].unique()
)
print(
    f"\nUnique raw times for inferred pair {selected_pair} "
    f"({len(inferred_pair_times)} values):"
)
print(inferred_pair_times)
```
